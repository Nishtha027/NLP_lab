```python
# Install/import required libraries and download NLTK resources
!pip -q install nltk pandas

import nltk
import pandas as pd
from nltk.corpus import wordnet as wn

nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)
nltk.download('brown', quiet=True)
nltk.download('punkt', quiet=True)

print('NLTK resources downloaded successfully.')
print('WordNet synsets available:', len(list(wn.all_synsets())))
```

```python
# Q1(a–d): Display selected WordNet synsets, definitions and examples

words = ['car', 'bank', 'bat', 'dog', 'light', 'plant', 'book', 'teacher', 'student', 'computer']

q1_rows = []

for word in words:
    synsets = wn.synsets(word)
    print('=' * 80)
    print(f'WORD: {word.upper()}')
    print(f'Number of synsets: {len(synsets)}')

    # Display all available synsets for each word
    for i, syn in enumerate(synsets, start=1):
        examples = '; '.join(syn.examples()) if syn.examples() else 'No example available'
        print(f'\n{i}. Synset: {syn.name()}')
        print(f'   Definition: {syn.definition()}')
        print(f'   Example: {examples}')

        q1_rows.append({
            'Word': word,
            'Synset': syn.name(),
            'Definition': syn.definition(),
            'Example': examples
        })

q1_df = pd.DataFrame(q1_rows)

print('\nQ1 Summary DataFrame:')
display(q1_df)
```

```python
# Q1(e): Identify words having multiple synsets and briefly discuss their meanings

multi_meaning_words = ['bank', 'bat', 'light']

for word in multi_meaning_words:
    print(f'\n{word.upper()} – different meanings:')
    for syn in wn.synsets(word):
        print(f'- {syn.name()}: {syn.definition()}')

print('\nObservations:')
print('1. bank can refer to a financial institution or the side of a river.')
print('2. bat can refer to a flying mammal or a piece of sports equipment.')
print('3. light can refer to illumination or something that is not heavy, among other senses.')
print('WordNet represents these different meanings as separate synsets.')
```

# Q2. Hypernyms and Hierarchical Semantic Relationships

A **hypernym** represents a more general category. For example, a dog is an animal, so an appropriate WordNet chain can move from `dog` toward more general concepts such as `canine → carnivore → mammal → ...` depending on the selected synset.

```python
# Q2(a–c): Select at least 10 nouns and extract direct hypernyms

noun_words = ['dog', 'cat', 'car', 'computer', 'teacher', 'student', 'book', 'tree', 'rose', 'airplane']

def first_noun_synset(word):
    """Return the first noun synset for a word, if available."""
    noun_synsets = wn.synsets(word, pos=wn.NOUN)
    return noun_synsets[0] if noun_synsets else None

q2_rows = []

for word in noun_words:
    syn = first_noun_synset(word)
    if syn is None:
        q2_rows.append({'Word': word, 'Synset': 'Not found', 'Direct Hypernyms': 'Not found'})
        continue

    direct = syn.hypernyms()
    direct_names = ', '.join(h.name() for h in direct) if direct else 'None'

    q2_rows.append({
        'Word': word,
        'Synset': syn.name(),
        'Direct Hypernyms': direct_names
    })

q2_df = pd.DataFrame(q2_rows)
display(q2_df)
```

```python
# Q2(d): Trace the hypernym hierarchy for at least five words (two or more levels)

def hypernym_chain(synset, max_levels=6):
    """Follow the first hypernym branch for a readable multi-level hierarchy."""
    chain = [synset.name()]
    current = synset

    for _ in range(max_levels):
        hypers = current.hypernyms()
        if not hypers:
            break
        current = hypers[0]
        chain.append(current.name())
    return ' → '.join(chain)

hierarchy_words = ['dog', 'cat', 'car', 'tree', 'rose', 'airplane']
hierarchy_rows = []

for word in hierarchy_words:
    syn = first_noun_synset(word)
    if syn:
        hierarchy_rows.append({
            'Word': word,
            'Synset': syn.name(),
            'Hypernym Chain (multiple levels)': hypernym_chain(syn, max_levels=6)
        })

q2_hierarchy_df = pd.DataFrame(hierarchy_rows)
display(q2_hierarchy_df)

print('Observation: moving rightward in a hypernym chain generally moves from a specific concept to increasingly general concepts.')
```

```python
# Q3(a–e): Calculate path similarity for at least eight word pairs

word_pairs = [
    ('car', 'automobile'),
    ('dog', 'cat'),
    ('dog', 'animal'),
    ('dog', 'computer'),
    ('car', 'bus'),
    ('teacher', 'student'),
    ('book', 'magazine'),
    ('rose', 'flower'),
    ('airplane', 'vehicle')
]

def best_noun_similarity(word1, word2):
    """Find the highest noun-synset path similarity between two words."""
    synsets1 = wn.synsets(word1, pos=wn.NOUN)
    synsets2 = wn.synsets(word2, pos=wn.NOUN)

    if not synsets1 or not synsets2:
        return None, None, None

    best_score = -1
    best_pair = (None, None)

    for s1 in synsets1:
        for s2 in synsets2:
            score = s1.path_similarity(s2)
            if score is not None and score > best_score:
                best_score = score
                best_pair = (s1, s2)

    if best_score == -1:
        return None, best_pair[0], best_pair[1]

    return best_score, best_pair[0], best_pair[1]

q3_rows = []

for w1, w2 in word_pairs:
    score, syn1, syn2 = best_noun_similarity(w1, w2)
    q3_rows.append({
        'Word 1': w1,
        'Word 2': w2,
        'Selected Synset 1': syn1.name() if syn1 else 'Not found',
        'Selected Synset 2': syn2.name() if syn2 else 'Not found',
        'Path Similarity': score
    })

q3_df = pd.DataFrame(q3_rows)
q3_ranked_df = q3_df.sort_values('Path Similarity', ascending=False, na_position='last').reset_index(drop=True)
q3_ranked_df.insert(0, 'Rank', range(1, len(q3_ranked_df) + 1))

display(q3_ranked_df)
```

```python
# Q3(f): Explain why different pairs receive different similarity values

print('Q3 Observations:')
print('- car–automobile is expected to receive a very high score because the words are near-synonyms and have closely related WordNet concepts.')
print('- dog–cat is highly related because both belong to closely connected animal categories.')
print('- dog–animal is related because animal is a general category (hypernym) of dog.')
print('- dog–computer is expected to be much less similar because they belong to very different semantic categories.')
print('- The exact scores depend on the synsets selected and the structure of the WordNet hierarchy.')
```
